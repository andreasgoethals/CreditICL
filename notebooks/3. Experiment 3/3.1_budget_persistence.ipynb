## B · Speed-up or persistent gain?
The horizontal axis is cumulative synthetic tasks (64 per update); mark 800,000 and 12.8 million

at common updates. A short control is never extrapolated to a longer treatment. The final
designated-holdout comparison is separate, after complete five-fold benchmarking.

```python
if paper.budget_selected("pd") or paper.budget_selected("lgd"):
    FIGS.save(paper.budget_gain("pd"), "pd_gain_by_budget",
        caption="PD development-score difference against equally trained control per matched seed against cumulative synthetic tasks, with the 0.8-million and 12.8-million task budgets marked; only common observed updates are compared.");
    FIGS.save(paper.budget_gain("lgd"), "lgd_gain_by_budget",
        caption="LGD development-score difference against equally trained control per matched seed against cumulative synthetic tasks, marking the 0.8-million and 12.8-million task budgets; no incomplete control trace is extrapolated.");
else:
    FIGS.save(campaign.availability("Both Experiment 3 mixtures await the documented development selection. No long-budget curves or holdout contrasts are available yet."), "budget_evidence_availability",
        caption="Availability notice for the frozen selected-versus-control long-budget comparison, two paired seeds per task and 12.8 million synthetic tasks per run.");
```

## C · Final designated-holdout contrast

```python
if paper.budget_selected("pd") and paper.final_ready("pd", "exp3"):
    FIGS.save(paper.paired_scores("pd", "exp3"), "pd_final_long_budget",
        caption="Final designated-holdout PD metric differences at 12.8 million tasks against the equally trained 0% control, paired over the two training seeds and five folds, with descriptive dataset bootstrap intervals.");
if paper.budget_selected("lgd") and paper.final_ready("lgd", "exp3"):
    FIGS.save(paper.distributional_scorecard("exp3"), "lgd_final_long_budget",
        caption="Final designated-holdout LGD R², approximate CRPS and endpoint-mass-error differences at 12.8 million tasks against the equally trained control, paired over seeds and folds with dataset-level intervals.");
```
